## DATA CLEANING

```python
clean_df = df.copy()                   # preserving the OG raw file/data#frame (df)
```

```python
clean_df
```

```python
# Converting "timestamp's" datatype to datetime and "Squawk's" datatype to String

clean_df["timestamp"]= pd.to_datetime(clean_df["timestamp"],utc =True) 

clean_df["timestamp"].dtype                    # to verify


clean_df["squawk"] = clean_df["squawk"].astype("string")

clean_df["squawk"].dtype                       # to verify

              
                         
```

```python
clean_df["airline"] = clean_df["airline"].replace({
     "indigo": "IndiGo",
    "AIR INDIA": "Air India",
    "AKASA AIR": "Akasa Air",
    "SPICEJET": "SpiceJet" 
})
```

```python
# to verify
clean_df["airline"].value_counts()            # now we can see, those values have been merged in their original categorical names
```

```python
clean_df= clean_df.drop_duplicates()              # to remove duplicates

# to verify
clean_df.duplicated().sum()
clean_df.shape
```

```python
# WE r going to avoid filling 0 in null values in both "latitude" & "longitude" bcz other flight observations may still be useful. 

clean_df.loc[                                               
    (clean_df["latitude"] < -90) |                       #this Finds any latitude outside -90 to +90 and replace it with a missing value(Nan).
    (clean_df["latitude"] > 90),
    "latitude"
] = np.nan                                              

# to verify
clean_df["latitude"].min()
clean_df["latitude"].max()
```

```python
clean_df.loc[
    (clean_df["longitude"] < -180) |
    (clean_df["longitude"] > 180),
    "longitude"
] = np.nan

# to verify
clean_df["longitude"].describe()
```

```python
clean_df = clean_df.copy()

clean_df["date"] = clean_df["timestamp"].dt.date                        # creating date, hour and day of week as asked in the instructions pdf
clean_df["hour"] = clean_df["timestamp"].dt.hour
clean_df["day_of_week"] = clean_df["timestamp"].dt.day_name()
```

*Flight_summary*

```python
# here we'll make a new dataframe as "flight_summary" for submission as aksed in the pdf's instruction

flight_summary = (
    clean_df.groupby("flight_number")                           #for each flight_number, these stats r created.
    .agg(
        first_seen=("timestamp", "min"),
        last_seen=("timestamp", "max"),
        observation_count=("timestamp", "count"),
        max_altitude=("altitude_ft", "max"),
        average_altitude=("altitude_ft", "mean"),
        average_speed=("ground_speed_kts", "mean")
    )
    .reset_index()
)
```

```python
flight_summary["tracking_duration"] = (
    flight_summary["last_seen"]                                           #tracking_duration (column) is added in the dataframe here
    - flight_summary["first_seen"]
)           
```

```python
flight_airline = (
    clean_df[["flight_number", "airline"]]
    .drop_duplicates()
)
                                                                   # first attached airline to each flight
flight_summary = flight_summary.merge(
    flight_airline,
    on="flight_number",
    how="left"
)
```

*Route ANALYTICS*

```python
# Remove records with missing origin or destination
route_df = clean_df.dropna(
    subset=["origin", "destination"]
).copy()

# Create route key
route_df["route"] = (
    route_df["origin"].astype(str)
    + "-"
    + route_df["destination"].astype(str)
)

# Creating  route summary
route_summary = (
    route_df.groupby("route")
    .agg(
        observed_flights=("flight_number", "nunique"),
        airlines=("airline", "nunique")
    )
    .reset_index()
    .sort_values(
        "observed_flights",
        ascending=False
    )
)

# Top 20 routes
top_20_routes = route_summary.head(20)

top_20_routes
```

**9. TELEMETRY QUALITY INVESTIGATION**

```python
telemetry_df = clean_df.copy()

# Invalid coordinate flags
telemetry_df["invalid_latitude_flag"] = (
    (telemetry_df["latitude"] < -90) |
    (telemetry_df["latitude"] > 90)
)

telemetry_df["invalid_longitude_flag"] = (
    (telemetry_df["longitude"] < -180) |
    (telemetry_df["longitude"] > 180)
)


# while Altitude, ground_speed and vertical_speed are in suspicious category due to its outlier
```

**FLIGHT-STATE ANALYSIS**

```python
state_df = clean_df.copy()

conditions = [
    
    # GROUND
    (
        state_df["on_ground"].fillna(False)
        |
        (
            (state_df["altitude_ft"] <= 1000)
            &
            (state_df["ground_speed_kts"] <= 80)
        )
    ),

    # CLIMBING
    (
        (~state_df["on_ground"].fillna(False))
        &
        (state_df["vertical_speed_fpm"] > 300)
        &
        (state_df["altitude_ft"] > 1000)
    ),

    # DESCENDING
    (
        (~state_df["on_ground"].fillna(False))
        &
        (state_df["vertical_speed_fpm"] < -300)
        &
        (state_df["altitude_ft"] > 1000)
    )
]

choices = [
    "GROUND",
    "CLIMBING",
    "DESCENDING"
]

state_df["flight_state"] = np.select(
    conditions,
    choices,
    default="CRUISING"
)
```

```python
# State distribution

state_summary = (
    state_df["flight_state"]
    .value_counts()
    .rename_axis("flight_state")
    .reset_index(name="observations")
)


state_summary["percentage"] = (
    state_summary["observations"]
    / len(state_df)
    * 100
).round(2)


state_summary
```

```python
state_summary.to_csv(
    "flight_state_summary.csv",
    index=False
)                                                      #saved the result
```

```python
# CALCULATING state distribution OR  compare states by airline or aircraft type**


state_by_airline = pd.crosstab(
    state_df["airline"],
    state_df["flight_state"],
    normalize="index"
) * 100

state_by_airline = state_by_airline.round(2)

state_by_airline
```
